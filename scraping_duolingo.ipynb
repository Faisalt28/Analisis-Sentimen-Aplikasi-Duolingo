```python
from google_play_scraper import reviews, Sort
import pandas as pd
import time
```

```python
app_id = 'com.duolingo'
all_reviews = []
next_token = None

print("Mulai scraping...")

while True:
    result, next_token = reviews(
        app_id,
        lang='en',
        country='us',
        sort=Sort.NEWEST,
        count=200,
        continuation_token=next_token
    )

    all_reviews.extend(result)
    print(f"Total sementara: {len(all_reviews)}")

    if not next_token or len(all_reviews) >= 30000:
        break

    time.sleep(1)

print("Scraping selesai!")

df = pd.DataFrame(all_reviews)
df = df[['userName', 'score', 'content', 'at']]
df.drop_duplicates(subset=['userName','content','at'], inplace=True)

print("Total setelah drop duplicate:", len(df))

df.to_csv("duolingo_reviews.csv", index=False)
print("Dataset disimpan sebagai duolingo_reviews.csv")
```
